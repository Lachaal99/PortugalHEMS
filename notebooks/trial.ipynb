```python
def get_project_root() -> Path:
    """Get the root directory of the project."""
    return Path.cwd().parent

root = get_project_root()
filepath = root / 'data' / 'raw' / 'PV Generation House 1.csv'
df_pv = pd.read_csv(filepath)
df_pv['Timestamp'] = pd.to_datetime(df_pv['Timestamp'], format='%d/%m/%Y %H:%M')
filepath  = root / 'data' / 'raw' / 'Load House 1.csv'
df_load = pd.read_csv(filepath)
df_load['DateTime'] = pd.to_datetime(df_load['DateTime'], format='%d/%m/%Y %H:%M')



```
